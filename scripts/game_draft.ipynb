```python
#Proyecto juego

def Instrucciones():
    # Menu inicial y controles
    print('''Calabozo de mis penas - Juego de rol
    ______________________________
    Controles:
        ir[dirección] Ej : norte, sur , este, oeste
        recoger[objeto]
        ______________________________
    ''')

inventario=[] 
sala_actual="El Calabozo"
salas={
    "El Calabozo" :{
        "norte": "El Pasillo",
        "objeto" : "machete"
    },
    "El Pasillo" : {
        "sur": "El Calabozo",
        "oeste": "La Torre",
        "objeto" : "aguardiente"
    },
    "La Torre" : {
        "este": "El Pasillo",
        "sur":"El Monte",
        "objeto":"llave"
    },
    "El Monte" : {
        "norte" : "La Torre",
        "objeto" : "Patasola"   
    }
}

print("Estás en", sala_actual)

Instrucciones()

#Acciones de navegacion 
while True:
    movimiento=input(">")
    print("DEBUG raw input:", repr(movimiento), flush=True)
    movimiento=movimiento.split(" ", 1)
    print("DEBUG parsed:", movimiento, "   sala_actual antes:", sala_actual, flush=True)
    if movimiento[0]=="ir":
        if movimiento[1] in salas[sala_actual]:
            sala_actual=salas[sala_actual][movimiento[1]]
            print(f"Estás en {sala_actual}.")
        else: print(f"{movimiento[1]} es inaccesible desde donde estás.")
```

```python
#Acciones de inventario
while True:
    movimiento=input(">")
    movimiento=movimiento.split(" ", 1)
    if movimiento[0] =="recoger":
        if movimiento[1] ==salas[sala_actual]["objeto"]:
            print(f"Tienes el objeto {movimiento[1]}.")
            inventario.append(movimiento[1])
            salas[sala_actual]["objeto"]="" 
        else: print(f"No hay {movimiento[1]}" "aquí.")

#Condiciones de victoria 1 : Escapar
if "llave" in inventario and "aguardiente" in inventario:
    print("¡Escapaste del calabozo con la llave y el aguardiente!")
    break

#Condicion de derrota :
```
